```python
from pdf2image import convert_from_path
from PIL import Image
from fpdf import FPDF
import os

# 1️⃣ Configuración inicial
# Ruta del archivo PDF que deseas procesar
pdf_file = "graficas_boxplots.pdf"

# Crear carpeta para guardar las imágenes temporalmente
output_folder = "pdf_images"
os.makedirs(output_folder, exist_ok=True)

# 2️⃣ Convertir PDF a imágenes
print("Convirtiendo cada página del PDF en imágenes...")
images = convert_from_path(pdf_file, dpi=150)  # dpi ajusta la calidad
image_paths = []

# Guardar cada página como una imagen
for i, page in enumerate(images):
    image_path = os.path.join(output_folder, f"page_{i+1}.png")
    page.save(image_path, "PNG")
    image_paths.append(image_path)
    print(f"Página {i+1} guardada como imagen: {image_path}")

# 3️⃣ Reconstruir un nuevo PDF con las imágenes
output_pdf = "nuevo_pdf_boxplots.pdf"
pdf = FPDF()

print("Reconstruyendo el PDF a partir de las imágenes...")
for img_path in image_paths:
    img = Image.open(img_path)
    width, height = img.size
    pdf.add_page()
    # Convertir píxeles a milímetros para FPDF
    pdf_w = 210  # Tamaño A4 por defecto en mm
    pdf_h = (height / width) * pdf_w
    pdf.image(img_path, 0, 0, pdf_w, pdf_h)

# Guardar el nuevo PDF
pdf.output(output_pdf)
print(f"Nuevo PDF creado: {output_pdf}")
```

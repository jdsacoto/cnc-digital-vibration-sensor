```python
import pandas as pd

# Función para leer un archivo CSV y cargarlo en un DataFrame de Pandas
def load_csv_to_dataframe(file_path):
    """
    Carga un archivo CSV en un DataFrame de Pandas.
    
    Parámetros:
    - file_path: str, ruta del archivo CSV
    
    Retorna:
    - df: DataFrame de Pandas con los datos cargados
    """
    try:
        df = pd.read_csv(file_path)
        print(f"✅ Archivo cargado exitosamente: {file_path}")
        print(f"📊 Tamaño del DataFrame: {df.shape}")
        return df
    except Exception as e:
        print(f"❌ Error al cargar el archivo: {e}")
        return None

# Ruta del archivo (puedes cambiarla según tu entorno)
file_path = "df_final.csv"

# Cargar el DataFrame
df = load_csv_to_dataframe(file_path)

# Mostrar las primeras filas
df.head()



```

```python
from sklearn.model_selection import train_test_split

# Define target variables
target_variables = ["ACCEL_PEAK", "ACCEL_RMS", "ACCEL_RMS_FREQ"]

# Step 1: Select the first target variable
target_variable = "ACCEL_PEAK"  # Start with ACCEL_PEAK

# Separate features (X) and target (y)
X = df.drop(columns=target_variables)  # Keep only input features
y = df[target_variable]  # Select the current target

# Split into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Print dataset sizes
print(f"✅ Training Set: {X_train.shape[0]} samples, {X_train.shape[1]} features")
print(f"✅ Test Set: {X_test.shape[0]} samples, {X_test.shape[1]} features")
print(f"🎯 Current Target Variable: {target_variable}")
```

# accel_peak

```python
import pandas as pd
import joblib
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, cross_val_score, train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error


# 📌 **Paso 1: Entrenar el modelo RandomForest**
print(f"\n🚀 Training Random Forest Model for {target_variable}...")

rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

# 📌 **Evaluar el modelo**
y_pred = rf_model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print(f"\n📊 {target_variable} - Basic Model Metrics:")
print(f"   🟢 Mean Squared Error (MSE): {mse:.4f}")
print(f"   🟡 Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"   🔵 R² Score: {r2:.4f}")
print(f"   🔴 Mean Absolute Error (MAE): {mae:.4f}")

# 📌 **Paso 2: Optimización de Hiperparámetros con GridSearchCV**
print("\n🔍 Optimizing Hyperparameters with GridSearchCV...")

param_grid = {
    "n_estimators": [300, 500, 700],  
    "max_depth": [None, 20, 50],  
    "min_samples_split": [2, 5, 10],  
    "min_samples_leaf": [1, 2, 4, 6],  
}

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42, n_jobs=-1),
    param_grid=param_grid,
    cv=3,  # 3-fold cross-validation
    n_jobs=-1,
    verbose=2
)

# **Entrenar GridSearch**
grid_search.fit(X_train, y_train)

# **Obtener el mejor modelo**
best_rf_model = grid_search.best_estimator_

# 📌 **Guardar el mejor modelo**
best_model_filename = f"random_forest_best_model_{target_variable}.pkl"
joblib.dump(best_rf_model, best_model_filename)
print(f"💾 Mejor modelo guardado en '{best_model_filename}'")

# 📌 **Evaluar el mejor modelo**
y_pred_best = best_rf_model.predict(X_test)

best_mse = mean_squared_error(y_test, y_pred_best)
best_rmse = np.sqrt(best_mse)
best_r2 = r2_score(y_test, y_pred_best)
best_mae = mean_absolute_error(y_test, y_pred_best)

print(f"\n🏆 {target_variable} - Tuned Model Metrics:")
print(f"   🟢 Best Model MSE: {best_mse:.4f}")
print(f"   🟡 Best Model RMSE: {best_rmse:.4f}")
print(f"   🔵 Best Model R² Score: {best_r2:.4f}")
print(f"   🔴 Best Model MAE: {best_mae:.4f}")

print(f"✅ Best Hyperparameters: {grid_search.best_params_}")

# 📌 **Paso 3: Validación del Modelo con Cross-Validation**
print("\n🛠 Validating Model with Cross-Validation...")

cv_r2_scores = cross_val_score(best_rf_model, X_train, y_train, cv=5, scoring="r2")
cv_mae_scores = cross_val_score(best_rf_model, X_train, y_train, cv=5, scoring="neg_mean_absolute_error")

print(f"📊 {target_variable} - Cross-Validation R² Scores: {cv_r2_scores}")
print(f"📈 Average Cross-Validation R² Score: {cv_r2_scores.mean():.4f}")

print(f"📊 {target_variable} - Cross-Validation MAE Scores: {-cv_mae_scores}")  
print(f"📈 Average Cross-Validation MAE Score: {-cv_mae_scores.mean():.4f}")

```

```python
import joblib
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.model_selection import GridSearchCV, cross_val_score
from sklearn.metrics import mean_squared_error, r2_score

# 🚀 Entrenar Modelo Multi-Output Random Forest
print("\n🚀 Training Enhanced Multi-Output Random Forest Model...")

# Modelo base con parámetros iniciales
multi_rf_model = MultiOutputRegressor(RandomForestRegressor(
    n_estimators=500,  # Mayor estabilidad
    max_depth=None,  # Permitir árboles completos
    min_samples_split=5,  # Evitar splits pequeños
    min_samples_leaf=3,  # Evitar hojas muy pequeñas
    max_features="sqrt",  # Reducir overfitting
    random_state=42,
    n_jobs=-1
))
multi_rf_model.fit(X_train, Y_train)

# Guardar modelo base
joblib.dump(multi_rf_model, "multi_rf_model.pkl")
print("💾 Model saved as 'multi_rf_model.pkl'")

# 📊 Evaluación Inicial
y_multi_pred = multi_rf_model.predict(X_test)
for i, target_variable in enumerate(target_variables):
    mse = mean_squared_error(Y_test[target_variable], y_multi_pred[:, i])
    r2 = r2_score(Y_test[target_variable], y_multi_pred[:, i])
    print(f"\n📊 {target_variable} - Enhanced Model MSE: {mse:.4f}")
    print(f"📈 {target_variable} - Enhanced Model R² Score: {r2:.4f}")

# 🔍 Optimización de Hiperparámetros con GridSearchCV
print("\n🔍 Optimizing Hyperparameters with GridSearchCV...")
param_grid = {
    "n_estimators": [500, 700],
    "max_depth": [20, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [2, 3, 5],
    "max_features": ["sqrt", "log2"]
}

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42, n_jobs=-1),
    param_grid=param_grid,
    cv=3,
    n_jobs=-1,
    verbose=2
)
grid_search.fit(X_train, Y_train[target_variables[0]])  # Tune con un target

# Modelo optimizado
best_rf_regressor = grid_search.best_estimator_
best_multi_rf_model = MultiOutputRegressor(best_rf_regressor)
best_multi_rf_model.fit(X_train, Y_train)

# Guardar modelo optimizado
joblib.dump(best_multi_rf_model, "best_multi_rf_model.pkl")
print("💾 Optimized Model saved as 'best_multi_rf_model.pkl'")

# 📊 Evaluación del Mejor Modelo
y_multi_pred_best = best_multi_rf_model.predict(X_test)
for i, target_variable in enumerate(target_variables):
    best_mse = mean_squared_error(Y_test[target_variable], y_multi_pred_best[:, i])
    best_r2 = r2_score(Y_test[target_variable], y_multi_pred_best[:, i])
    print(f"\n🏆 {target_variable} - Best Model MSE: {best_mse:.4f}")
    print(f"🏆 {target_variable} - Best Model R² Score: {best_r2:.4f}")

print(f"✅ Best Hyperparameters: {grid_search.best_params_}")

# 🛠 Validación del Modelo con Cross-Validation
print("\n🛠 Validating Model with Cross-Validation...")
for i, target_variable in enumerate(target_variables):
    cv_scores = cross_val_score(best_rf_regressor, X_train, Y_train[target_variable], cv=5, scoring="r2")
    print(f"📊 {target_variable} - Cross-Validation R² Scores: {cv_scores}")
    print(f"📈 {target_variable} - Avg Cross-Validation R² Score: {cv_scores.mean():.4f}")
```

```python
import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt
import optuna
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler
from sklearn.multioutput import MultiOutputRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor, StackingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.neural_network import MLPRegressor
import tensorflow as tf
from tensorflow import keras

# ------------------------------
# 📂 1. Cargar y Preprocesar Datos
# ------------------------------
def load_csv_to_dataframe(file_path):
    df = pd.read_csv(file_path)
    df.dropna(inplace=True)  # Elimina valores nulos
    print(f"✅ Archivo cargado: {file_path}")
    return df

file_path = "df_final.csv"
df = load_csv_to_dataframe(file_path)

# ------------------------------
# 🔄 2. División de Datos
# ------------------------------
target_variables = ["ACCEL_PEAK", "ACCEL_RMS", "ACCEL_RMS_FREQ"]
X = df.drop(columns=target_variables)
Y = df[target_variables]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size=0.20, random_state=42)

# ------------------------------
# 🚀 3. Definir Modelos
# ------------------------------
models = {
    "RandomForest": MultiOutputRegressor(RandomForestRegressor(n_estimators=500, max_depth=50, min_samples_split=2, min_samples_leaf=2, random_state=42, n_jobs=-1)),
    "GradientBoosting": MultiOutputRegressor(GradientBoostingRegressor(n_estimators=300, learning_rate=0.05, max_depth=10, random_state=42)),
    "ExtraTrees": MultiOutputRegressor(ExtraTreesRegressor(n_estimators=500, max_depth=50, random_state=42, n_jobs=-1)),
    "XGBoost": MultiOutputRegressor(XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=10, random_state=42)),
    "LightGBM": MultiOutputRegressor(LGBMRegressor(n_estimators=300, learning_rate=0.05, max_depth=10, random_state=42)),
    "CatBoost": MultiOutputRegressor(CatBoostRegressor(iterations=300, learning_rate=0.05, depth=10, verbose=0, random_state=42)),
    "MLPRegressor": MultiOutputRegressor(MLPRegressor(hidden_layer_sizes=(128, 64, 32), activation='relu', solver='adam', max_iter=500, random_state=42))
}

# ------------------------------
# 🏆 4. Entrenamiento y Evaluación
# ------------------------------
results = {}

for model_name, model in models.items():
    print(f"\n🚀 Training {model_name} Model...")
    model.fit(X_train, Y_train)
    joblib.dump(model, f"{model_name.lower()}_multioutput.pkl")
    Y_pred = model.predict(X_test)
    mse = mean_squared_error(Y_test, Y_pred, multioutput='raw_values')
    rmse = np.sqrt(mse)
    r2 = r2_score(Y_test, Y_pred, multioutput='raw_values')
    mae = mean_absolute_error(Y_test, Y_pred, multioutput='raw_values')
    results[model_name] = {"MSE": mse, "RMSE": rmse, "R²": r2, "MAE": mae}

# ------------------------------
# 🔍 5. Optimización con Optuna
# ------------------------------
def objective(trial):
    n_estimators = trial.suggest_int("n_estimators", 100, 500)
    max_depth = trial.suggest_int("max_depth", 3, 50)
    learning_rate = trial.suggest_float("learning_rate", 0.01, 0.1)
    model = XGBRegressor(n_estimators=n_estimators, learning_rate=learning_rate, max_depth=max_depth, random_state=42)
    multi_output_model = MultiOutputRegressor(model)
    scores = cross_val_score(multi_output_model, X_train, Y_train, cv=3, scoring="r2")
    return np.mean(scores)

study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=20)
print("✅ Best Hyperparameters:", study.best_params)

# ------------------------------
# 🤖 6. Red Neuronal con Keras
# ------------------------------
def build_keras_model(input_shape):
    model = keras.Sequential([
        keras.layers.Dense(128, activation='relu', input_shape=(input_shape,)),
        keras.layers.Dense(64, activation='relu'),
        keras.layers.Dense(32, activation='relu'),
        keras.layers.Dense(len(target_variables))
    ])
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    return model

keras_model = build_keras_model(X_train.shape[1])
history = keras_model.fit(X_train, Y_train, epochs=50, batch_size=32, validation_data=(X_test, Y_test), verbose=1)
keras_model.save("keras_multioutput_model.h5")

# ------------------------------
# 📊 7. Comparación de Modelos
# ------------------------------
print("\n📌 Summary of Model Performance:")
for key, value in results.items():
    print(f"\n🔹 {key}:")
    for i, target_variable in enumerate(target_variables):
        print(f"   🎯 {target_variable}:")
        print(f"      🟢 MSE: {value['MSE'][i]:.4f}")
        print(f"      🟡 RMSE: {value['RMSE'][i]:.4f}")
        print(f"      🔵 R²: {value['R²'][i]:.4f}")
        print(f"      🔴 MAE: {value['MAE'][i]:.4f}")

print("✅ Modelos entrenados y evaluados exitosamente.")
```

```python
import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt
import optuna
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler
from sklearn.multioutput import MultiOutputRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor, StackingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.neural_network import MLPRegressor
import tensorflow as tf
from tensorflow import keras

# ------------------------------
# 📂 1. Cargar y Preprocesar Datos
# ------------------------------
def load_csv_to_dataframe(file_path):
    df = pd.read_csv(file_path)
    df.dropna(inplace=True)  # Elimina valores nulos
    print(f"✅ Archivo cargado: {file_path}")
    return df

file_path = "df_final.csv"
df = load_csv_to_dataframe(file_path)

# ------------------------------
# 🔄 2. División de Datos
# ------------------------------
target_variables = ["ACCEL_PEAK", "ACCEL_RMS", "ACCEL_RMS_FREQ"]
X = df.drop(columns=target_variables)
Y = df[target_variables]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size=0.20, random_state=42)

# ------------------------------
# 🚀 3. Definir Modelos
# ------------------------------
models = {
    "RandomForest": MultiOutputRegressor(RandomForestRegressor(n_estimators=600, max_depth=30, min_samples_split=4, min_samples_leaf=2, random_state=42, n_jobs=-1)),
    "GradientBoosting": MultiOutputRegressor(GradientBoostingRegressor(n_estimators=400, learning_rate=0.05, max_depth=12, random_state=42)),
    "ExtraTrees": MultiOutputRegressor(ExtraTreesRegressor(n_estimators=600, max_depth=30, random_state=42, n_jobs=-1)),
    "XGBoost": MultiOutputRegressor(XGBRegressor(n_estimators=400, learning_rate=0.03, max_depth=15, random_state=42)),
    "LightGBM": MultiOutputRegressor(LGBMRegressor(n_estimators=400, learning_rate=0.03, max_depth=12, random_state=42)),
    "CatBoost": MultiOutputRegressor(CatBoostRegressor(iterations=400, learning_rate=0.03, depth=12, verbose=0, random_state=42)),
    "MLPRegressor": MultiOutputRegressor(MLPRegressor(hidden_layer_sizes=(256, 128, 64), activation='relu', solver='adam', max_iter=800, random_state=42))
}

# ------------------------------
# 🏆 4. Entrenamiento y Evaluación
# ------------------------------
results = {}

for model_name, model in models.items():
    print(f"\n🚀 Training {model_name} Model...")
    model.fit(X_train, Y_train)
    joblib.dump(model, f"{model_name.lower()}_multioutput.pkl")
    Y_pred = model.predict(X_test)
    mse = mean_squared_error(Y_test, Y_pred, multioutput='raw_values')
    rmse = np.sqrt(mse)
    r2 = r2_score(Y_test, Y_pred, multioutput='raw_values')
    mae = mean_absolute_error(Y_test, Y_pred, multioutput='raw_values')
    results[model_name] = {"MSE": mse, "RMSE": rmse, "R²": r2, "MAE": mae}

# ------------------------------
# 🔍 5. Optimización con Optuna
# ------------------------------
def objective(trial):
    n_estimators = trial.suggest_int("n_estimators", 100, 600)
    max_depth = trial.suggest_int("max_depth", 5, 20)
    learning_rate = trial.suggest_float("learning_rate", 0.01, 0.1)
    model = XGBRegressor(n_estimators=n_estimators, learning_rate=learning_rate, max_depth=max_depth, random_state=42)
    multi_output_model = MultiOutputRegressor(model)
    scores = cross_val_score(multi_output_model, X_train, Y_train, cv=3, scoring="r2")
    return np.mean(scores)

study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=30)
print("✅ Best Hyperparameters:", study.best_params)

# ------------------------------
# 🤖 6. Red Neuronal con Keras
# ------------------------------
def build_keras_model(input_shape):
    model = keras.Sequential([
        keras.layers.Dense(256, activation='relu', input_shape=(input_shape,)),
        keras.layers.Dropout(0.3),
        keras.layers.Dense(128, activation='relu'),
        keras.layers.Dropout(0.3),
        keras.layers.Dense(64, activation='relu'),
        keras.layers.Dropout(0.2),
        keras.layers.Dense(len(target_variables))
    ])
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001), loss='mse', metrics=['mae'])
    return model

keras_model = build_keras_model(X_train.shape[1])
history = keras_model.fit(X_train, Y_train, epochs=100, batch_size=32, validation_data=(X_test, Y_test), verbose=1)
keras_model.save("keras_multioutput_model.keras")  # Nuevo formato en vez de .h5

# ------------------------------
# 📊 7. Comparación de Modelos
# ------------------------------
print("\n📌 Summary of Model Performance:")
for key, value in results.items():
    print(f"\n🔹 {key}:")
    for i, target_variable in enumerate(target_variables):
        print(f"   🎯 {target_variable}:")
        print(f"      🟢 MSE: {value['MSE'][i]:.4f}")
        print(f"      🟡 RMSE: {value['RMSE'][i]:.4f}")
        print(f"      🔵 R²: {value['R²'][i]:.4f}")
        print(f"      🔴 MAE: {value['MAE'][i]:.4f}")

print("✅ Modelos entrenados y evaluados exitosamente.")
```

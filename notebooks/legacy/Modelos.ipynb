```python
import pandas as pd

# Función para leer un archivo CSV y cargarlo en un DataFrame de Pandas
def load_csv_to_dataframe(file_path):
    """
    Carga un archivo CSV en un DataFrame de Pandas.
    
    Parámetros:
    - file_path: str, ruta del archivo CSV
    
    Retorna:
    - df: DataFrame de Pandas con los datos cargados
    """
    try:
        df = pd.read_csv(file_path)
        print(f"✅ Archivo cargado exitosamente: {file_path}")
        print(f"📊 Tamaño del DataFrame: {df.shape}")
        return df
    except Exception as e:
        print(f"❌ Error al cargar el archivo: {e}")
        return None

# Ruta del archivo (puedes cambiarla según tu entorno)
file_path = "df_final.csv"

# Cargar el DataFrame
df = load_csv_to_dataframe(file_path)

# Mostrar las primeras filas
df.head()



```

```python
df_zero_load = df.loc[217:247]  # Filtrar por rango de índices

# Mostrar el DataFrame resultante
print(df_zero_load)
```

```python
df_load = df.loc[1600:1680]  # Filtrar por rango de índices

# Mostrar el DataFrame resultante
print(df_load)
```

# RF ACCEL_PEAK NO LOAD SPINDLE

```python
import joblib
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# 📂 Load the model using joblib with error handling
try:
    model = joblib.load("Modelos/random_forest_best_model_ACCEL_PEAK.pkl")
    print("✅ Model loaded successfully!")
except Exception as e:
    print(f"❌ Error loading model: {e}")
    exit()

# 🏗️ Define target variables
target_variables = ["ACCEL_PEAK", "ACCEL_RMS", "ACCEL_RMS_FREQ"]
target_variable = "ACCEL_PEAK"  # Default target variable

# 🎯 Separate features (X) and target values (y)
if set(target_variables).issubset(df_zero_load.columns):
    X = df_zero_load.drop(columns=target_variables)  # Input features
    y_real = df_zero_load[target_variable]  # Actual values
else:
    print("❌ Error: Target variables are missing in df_zero_load.")
    exit()

# 🔮 Make predictions
y_pred = model.predict(X)

# 📊 Plot actual vs. predicted values
plt.figure(figsize=(10, 5))
plt.plot(y_real.values, label="Actual Values", marker='o', linestyle='-', alpha=0.7)
plt.plot(y_pred, label="Predicted Values", marker='s', linestyle='--', color='r', alpha=0.7)
plt.xlabel("Sample Index")
plt.ylabel(target_variable)
plt.title(f"Comparison: Actual vs. Predicted {target_variable}")
plt.legend()
plt.grid(True)
plt.show()

# 📉 Compute model evaluation metrics
mae = mean_absolute_error(y_real, y_pred)  # Mean Absolute Error
mse = mean_squared_error(y_real, y_pred)  # Mean Squared Error
rmse = np.sqrt(mse)  # Root Mean Squared Error
r2 = r2_score(y_real, y_pred)  # R² Score

# 📊 Print evaluation results
print("\n📊 Model Evaluation:")
print(f"✅ Mean Absolute Error (MAE): {mae:.4f}")
print(f"✅ Mean Squared Error (MSE): {mse:.4f}")
print(f"✅ Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"✅ R² Score: {r2:.4f}")

# 🔔 Quick Interpretation:
if r2 > 0.8:
    print("🚀 The model has high prediction accuracy.")
elif r2 > 0.5:
    print("📊 The model is acceptable but could improve.")
else:
    print("⚠️ The model is not predicting well, adjustments are needed.")

# 📊 Feature Importance (If the model supports it)
if hasattr(model, "feature_importances_"):
    importances = model.feature_importances_
    feature_names = X.columns

    plt.figure(figsize=(12, 6))
    plt.barh(feature_names, importances, color='skyblue')
    plt.xlabel("Feature Importance")
    plt.ylabel("Features")
    plt.title("Feature Importance from RandomForest Model")
    plt.grid(True)
    plt.show()
else:
    print("ℹ️ The model does not support feature importance extraction.")
```

```python
import joblib
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 📂 Load the model using joblib
model = joblib.load("Modelos/multi_rf_model.pkl")

# 🎯 Define target variables
target_variables = ["ACCEL_PEAK", "ACCEL_RMS", "ACCEL_RMS_FREQ"]

# 🔄 Separate features (X) and target values (y)
X = df_zero_load.drop(columns=target_variables)  # Input features
y_real = df_zero_load[target_variables]  # Actual values

# 🔮 Make predictions
y_pred = model.predict(X)

# ⏳ Create the time axis (each sample occurs every 0.2s)
time_axis = np.arange(len(y_real)) * 0.2  # Each sample occurs every 0.2 seconds

# 📊 Plot actual vs. predicted values separately for each variable
fig, axes = plt.subplots(nrows=len(target_variables), figsize=(10, 5 * len(target_variables)))

for i, variable in enumerate(target_variables):
    ax = axes[i] if len(target_variables) > 1 else axes  # Handle single variable case
    ax.plot(time_axis, y_real[variable], label="Actual Values", marker='o', linestyle='-')
    ax.plot(time_axis, y_pred[:, i], label="Predicted Values", marker='s', linestyle='--', color='r')
    ax.set_xlabel("Time (seconds)")
    ax.set_ylabel(variable)
    ax.set_title(f"Comparison of {variable}: Actual vs. Predicted")
    ax.legend()
    ax.grid(True)

plt.tight_layout()  # Adjust to prevent overlap
plt.show()

# 📈 MODEL EVALUATION METRICS FOR EACH VARIABLE
print("📊 Model Evaluation:")

for i, variable in enumerate(target_variables):
    mae = mean_absolute_error(y_real[variable], y_pred[:, i])  # Mean Absolute Error
    mse = mean_squared_error(y_real[variable], y_pred[:, i])  # Mean Squared Error
    rmse = np.sqrt(mse)  # Root Mean Squared Error
    r2 = r2_score(y_real[variable], y_pred[:, i])  # R² Score

    # 📊 Print metrics
    print(f"\n🔹 Evaluation for {variable}:")
    print(f"✅ MAE: {mae:.4f}")
    print(f"✅ MSE: {mse:.4f}")
    print(f"✅ RMSE: {rmse:.4f}")
    print(f"✅ R² Score: {r2:.4f}")

    # 🔔 Quick Interpretation:
    if r2 > 0.8:
        print("🚀 The model has high prediction accuracy.")
    elif r2 > 0.5:
        print("📊 The model is acceptable but could improve.")
    else:
        print("⚠️ The model is not predicting well, adjustments are needed.")
```

```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 📂 Cargar el modelo con pickle, con manejo de errores
try:
    with open("Modelos/xgb_model_accel_peak.pkl", "rb") as file:
        model = pickle.load(file)
    print("✅ Modelo cargado exitosamente.")
except Exception as e:
    print(f"❌ Error al cargar el modelo: {e}")
    exit()

# 🎯 Definir las variables objetivo
target_variables = ["ACCEL_PEAK", "ACCEL_RMS", "ACCEL_RMS_FREQ"]
target_variable = "ACCEL_PEAK"  # Variable objetivo

# 🔄 Separar características (X) y la variable objetivo (y)
if set(target_variables).issubset(df_zero_load.columns):
    X = df_zero_load.drop(columns=target_variables)  # Variables de entrada
    y_real = df_zero_load[target_variable]  # Valores reales
else:
    print("❌ Error: No se encontraron todas las variables objetivo en df_zero_load.")
    exit()

# 🔮 Hacer predicciones con el modelo cargado
y_pred = model.predict(X)

# ⏳ Crear el eje de tiempo (cada muestra ocurre cada 0.2s)
tiempo = np.arange(len(y_real)) * 0.2  # Cada muestra ocurre cada 0.2 segundos

# 📊 Graficar valores reales vs. valores pronosticados
plt.figure(figsize=(10, 5))
plt.plot(tiempo, y_real.values, label="Valores Reales", marker='o', linestyle='-', alpha=0.7)
plt.plot(tiempo, y_pred, label="Valores Pronosticados", marker='s', linestyle='--', color='r', alpha=0.7)
plt.xlabel("Tiempo (segundos)")
plt.ylabel(target_variable)
plt.title(f"Comparación de {target_variable}: Real vs. Pronosticado")
plt.legend()
plt.grid(True)
plt.show()

# 📉 Calcular métricas de evaluación del modelo
mae = mean_absolute_error(y_real, y_pred)  # Error absoluto medio
mse = mean_squared_error(y_real, y_pred)  # Error cuadrático medio
rmse = np.sqrt(mse)  # Raíz del MSE
r2 = r2_score(y_real, y_pred)  # Coeficiente de determinación

# 📊 Mostrar evaluación del modelo
print("\n📊 Evaluación del Modelo:")
print(f"✅ MAE: {mae:.4f}")
print(f"✅ MSE: {mse:.4f}")
print(f"✅ RMSE: {rmse:.4f}")
print(f"✅ R² Score: {r2:.4f}")

# 🔔 Interpretación rápida
if r2 > 0.8:
    print("🚀 El modelo tiene una alta precisión en la predicción.")
elif r2 > 0.5:
    print("📊 El modelo es aceptable pero puede mejorar.")
else:
    print("⚠️ El modelo no predice bien, ajustes son necesarios.")

# 📊 Mostrar importancia de características si el modelo lo permite
if hasattr(model, "feature_importances_"):
    importances = model.feature_importances_
    feature_names = X.columns

    plt.figure(figsize=(12, 6))
    plt.barh(feature_names, importances, color='skyblue')
    plt.xlabel("Importancia de la característica")
    plt.ylabel("Características")
    plt.title("Importancia de Características en XGBoost")
    plt.grid(True)
    plt.show()
else:
    print("ℹ️ El modelo no admite la extracción de importancia de características.")
```

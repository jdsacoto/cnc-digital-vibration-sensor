```python
import pandas as pd

# Función para leer un archivo CSV y cargarlo en un DataFrame de Pandas
def load_csv_to_dataframe(file_path):
    """
    Carga un archivo CSV en un DataFrame de Pandas.
    
    Parámetros:
    - file_path: str, ruta del archivo CSV
    
    Retorna:
    - df: DataFrame de Pandas con los datos cargados
    """
    try:
        df = pd.read_csv(file_path)
        print(f"✅ Archivo cargado exitosamente: {file_path}")
        print(f"📊 Tamaño del DataFrame: {df.shape}")
        return df
    except Exception as e:
        print(f"❌ Error al cargar el archivo: {e}")
        return None

# Ruta del archivo (puedes cambiarla según tu entorno)
file_path = "df_final.csv"

# Cargar el DataFrame
df = load_csv_to_dataframe(file_path)

# Mostrar las primeras filas
df.head()



```

```python
import joblib
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 📂 Cargar el modelo con joblib y manejo de errores
model_path = "Modelos2/randomforest_model_accel_peak.pkl"
try:
    model = joblib.load(model_path)
    print(f"✅ Modelo cargado exitosamente. Tipo: {type(model)}")
except Exception as e:
    print(f"❌ Error al cargar el modelo: {e}")
    exit()

# 🎯 Definir las variables objetivo
target_variables = ["ACCEL_PEAK", "ACCEL_RMS", "ACCEL_RMS_FREQ"]
target_variable = "ACCEL_PEAK"  # Variable objetivo

# 🔄 Verificar que las variables objetivo existan en el dataset
if set(target_variables).issubset(df_zero_load.columns):
    X = df_zero_load.drop(columns=target_variables)  # Variables de entrada
    y_real = df_zero_load[target_variable]  # Valores reales
else:
    print("❌ Error: No se encontraron todas las variables objetivo en df_zero_load.")
    exit()

# 🔮 Hacer predicciones con el modelo cargado
try:
    y_pred = model.predict(X)
    print(f"✅ Predicciones realizadas con éxito. Shape: {y_pred.shape}")
except Exception as e:
    print(f"❌ Error al realizar predicciones: {e}")
    exit()

# ⏳ Crear el eje de tiempo (cada muestra ocurre cada 0.2s)
tiempo = np.arange(len(y_real)) * 0.2  # Cada muestra ocurre cada 0.2 segundos

# 📊 Graficar valores reales vs. valores pronosticados
plt.figure(figsize=(10, 5))
plt.plot(tiempo, y_real.values, label="Valores Reales", marker='o', linestyle='-', alpha=0.7)
plt.plot(tiempo, y_pred, label="Valores Pronosticados", marker='s', linestyle='--', color='r', alpha=0.7)
plt.xlabel("Tiempo (segundos)")
plt.ylabel(target_variable)
plt.title(f"Comparación de {target_variable}: Real vs. Pronosticado")
plt.legend()
plt.grid(True)
plt.show()

# 📉 Calcular métricas de evaluación del modelo
mae = mean_absolute_error(y_real, y_pred)  # Error absoluto medio
mse = mean_squared_error(y_real, y_pred)  # Error cuadrático medio
rmse = np.sqrt(mse)  # Raíz del MSE
r2 = r2_score(y_real, y_pred)  # Coeficiente de determinación

# 📊 Mostrar evaluación del modelo
print("\n📊 Evaluación del Modelo:")
print(f"✅ MAE: {mae:.4f}")
print(f"✅ MSE: {mse:.4f}")
print(f"✅ RMSE: {rmse:.4f}")
print(f"✅ R² Score: {r2:.4f}")

# 🔔 Interpretación rápida
if r2 > 0.8:
    print("🚀 El modelo tiene una alta precisión en la predicción.")
elif r2 > 0.5:
    print("📊 El modelo es aceptable pero puede mejorar.")
else:
    print("⚠️ El modelo no predice bien, ajustes son necesarios.")

# 📊 Mostrar importancia de características si el modelo lo permite
if hasattr(model, "feature_importances_"):
    importances = model.feature_importances_
    feature_names = X.columns

    plt.figure(figsize=(12, 6))
    plt.barh(feature_names, importances, color='skyblue')
    plt.xlabel("Importancia de la característica")
    plt.ylabel("Características")
    plt.title("Importancia de Características en RandomForest")
    plt.grid(True)
    plt.show()
else:
    print("ℹ️ El modelo no admite la extracción de importancia de características.")
```
